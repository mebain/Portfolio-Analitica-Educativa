### Análisis realizado
- Exploración inicial del dataset (EDA)

### Información sobre el dataset

El dataset contiene información sobre el rendimiento académico de estudiantes en Matemática y Portugués, incluyendo variables sociodemográficas y académicas.

Se utilizaron dos archivos correspondientes a cada asignatura, los cuales fueron integrados para su análisis.

### Carga de datos

```python
#Leer archivo .csv con resultados de Matemática
ruta = os.path.join(".", "data","student-mat.csv") #construye ruta de archivos compatible con diferentes sistemas operativos.
df_math = pd.read_csv(ruta, sep=";") #Lee el archivo y lo guarda en un DataFrame de Pandas. 
```

```python
#Leer archivo .csv con resultados de Portugues
ruta = os.path.join(".", "data", "student-por.csv") #construye ruta de archivos compatible con diferentes sistemas operativos
df_port = pd.read_csv(ruta, sep=";") #Lee el archivo y lo guarda en un DataFrame de Pandas. 
```

### 📊 Análisis de datos

A continuación se realiza un análisis exploratorio para identificar patrones y relaciones entre las variables del dataset.

```python
# Se visualizan registros aleatorios del dataset para comprender su estructura.
display(df_math.sample(20)) #visualiza 20 filas aleatorias del dataframe, df_math. 
print()
display(df_port.sample(20)) #visualiza 20 filas aleatorias del dataframe df_port. 
```

```python
# Se visualizan los tipos de datos de cada Dataframe
print(f"Tipos de datos de los dataframes, MATEMÁTICA")
print(df_math.dtypes) # tipos de datos sobre las columnas
print()
print(f"Tipos de datos de los dataframes, PORTUGUES")
print (df_port.dtypes) # tipos de datos sobre las columnas
mask=(df_math.dtypes==df_port.dtypes).all() #devuelve True si las variables entre las dos tablas son iguales. 
print(mask)
```

```python
#Se obtiene información sobre la estructura, tipos de datos, valores nulos y consumo de memoria de cada DataFrame.
df_math.info()
df_port.info()
```

```python
print("Dataframe de Matemática")
display(df_math.describe(include='all')) #Se agrega a describe() include='all' para que tome todos los atributos, no sólo los numéricos. 
print()
print("Dataframe de Portugués")
display(df_port.describe(include='all')) #Se agrega a describe() include='all' para que tome todos los atributos, no sólo los numéricos. 
```

#### Nota adicional del dataset. 
Hay 382 estudiantes que pertenecen a ambas base de datos. Estos estudiantes pueden identificarse buscando
atributos idénticos que caracterizan a cada estudiante. Los atributos son: school,sex,age,address,famsize,Pstatus,Medu,Fedu,Mjob,Fjob,
reason,nursery,internet

```python
# Se combinan los dos datasets (df_math y df_port) y se muestra información del resultado: df_combined
df_combined= df_math.merge(df_port, on=['school','sex','age','address','famsize','Pstatus','Medu','Fedu','Mjob','Fjob','reason','nursery',
                                        'internet'], how='inner')
display(df_combined.head(5)) # me muestra los 5 primeros registros. 
print(f" La longitud del dataframe combinado es,{len(df_combined)}")
df_combined.info()
```

```python
print (f"Correlación de variables numéricas del DF Matemática\n")
math_corr= df_math.corr(numeric_only=True) # Correlación de variables sólo numéricas. 
np.fill_diagonal(math_corr.values, np.nan)
display (math_corr)
print (f"Correlación de variables numéricas del DF Portugues\n")
port_corr= df_port.corr(numeric_only=True) # Correlación de variables sólo numéricas. 
np.fill_diagonal(port_corr.values, np.nan)
display (port_corr)
```

### Sobre correlación
De lo anterior se puede ver una correlación próxima a cero en la mayoria de las variables, a excepción de los resultados (G1,G2 y G3) donde siempre es superior al 0.8 lo cual da cuenta de lo relacionados que están los resultados alcanzados en los exámenes parciales con el examen final.

## Visualización

```python
# magic command para visualizar gráficos en jupyter en modo no interactivo
%matplotlib inline
```

```python
#Toma cada Dataset y se fija en qué medida las ausencias impactan en las calificaciones. 
y_data = df_math['G3']
x_data = df_math['absences']
y1_data = df_port['G3']
x1_data = df_port['absences']

fig,axis = plt.subplots(2,1) #crea una figura con dos gráficos verticales. 

# Matemáticas
axis[0].scatter(x_data,y_data,label='Matemáticas', alpha=0.5) # crea gráfico de puntos
axis[0].legend(loc='best')
axis[0].set_title('Resultado final vs Ausencias- Matemáticas')
axis[0].set_xlabel("Ausencias")
axis[0].set_ylabel("Nota final (G3)")
axis[0].set(ylim=(0, 20))
axis[0].set_xlim([0,60])
axis[0].grid()
# Portugues
axis[1].scatter(x1_data,y1_data,label='Portugues',alpha=0.5, color='#F54927') # crea gráfico de puntos. Aqui con un color diferente.
axis[1].legend(loc='best')
axis[1].set_title('Resultado final vs Ausencias- Portugues')
axis[1].set_ylabel("Nota final (G3)")
axis[1].set(ylim=(0, 20))
axis[1].set_xlim([0,60])
axis[1].grid()

fig.tight_layout()
plt.show()
plt.close(fig) 
```

### Interpretación del gráfico
Como puede observarse en los gráficos, no se identifica una relación clara entre la cantidad de ausencias y el rendimiento académico final. 
Los datos presentan alta dispersión, lo que sugiere que las ausencias por sí solas no explican el desempeño de los estudiantes.

```python
#En este gráfico boxplot se fija en que medida el género impacta en los resultados. Trabaja con los dataframe por separado. 
fig, ax = plt.subplots(2,1, figsize=(6,8), sharey=True) #sharey: comparte el eje y. 
sns.boxplot(x='sex', y='G3', data=df_math, ax=ax[0], fill=False)
ax[0].set_title("Resultado final Matemática vs Género") #se setea el título del gráfico
ax[0].set(xlabel= 'Género', ylabel='Resultado final') #se asignan nombres a las columnas
ax[0].set_xticks([0,1])
ax[0].set_xticklabels(["Femenino","Masculino"])

sns.boxplot(x='sex', y='G3', data=df_port, ax=ax[1],color='#F54927', fill=False)
ax[1].set_title ("Resultado final Portugués vs Género") #se setea el título del gráfico
ax[1].set(xlabel= 'Género', ylabel='Resultado final') #se asignan nombres a las columnas
ax[1].set_xticks([0,1])
ax[1].set_xticklabels(["Femenino","Masculino"])

plt.tight_layout()

plt.show()
plt.close(fig) 
```

#### Interpretación del gráfico
Aunque visualmente parece observarse una ligera diferencia en el rendimiento según género, esta no puede considerarse concluyente sin un análisis estadístico adicional. 
Por lo tanto, no se puede afirmar que exista una diferencia significativa entre ambos grupos.

```python
#Este gráfico muestra la relación que existe entre los resultados parciales (G1 y G2) y la nota final (G3) para cada una de las materias 
#(dataframes). Permite ver si hay coherencia entre parciales y nota final. De igual manera, si alumnos con bajos G1 y G2 pueden igual tener 
#buen desempeño en G3, es decir la calificación final.

fig, ax = plt.subplots(1,2, figsize=(12,6),sharey=True)
sns.scatterplot(x='G1', y='G2', hue='G3', data=df_math, palette='viridis',ax=ax[0])
sns.scatterplot(x='G1', y='G2', hue='G3', data=df_port, palette='viridis',ax=ax[1])
ax[0].set_title("Matemática")
ax[0].set(xlabel= 'Resultados primer parcial', ylabel='Resultado segundo parcial')
ax[0].set_xlim([0,20])

ax[1].set_title("Portugués")
ax[1].set(xlabel= 'Resultados primer parcial', ylabel='Resultado segundo parcial')
ax[1].set_xlim([0,20])
plt.show()
plt.close(fig) 
```

```python
# Resultados finales alcanzados en función de la escuela.
# plot:
fig, ax = plt.subplots(1,2, figsize=(10,4), sharey=True) #sharey: comparte el eje y. 

sns.boxplot(x='school', y='G3_x', data=df_combined, ax=ax[0], fill=False)
ax[0].set_title("Resultado final Matemática vs Escuela") #se setea el título del gráfico
ax[0].set(xlabel= 'Escuela', ylabel='Resultado final') #se asignan nombres a las columnas
ax[0].set_xticks([0,1])
ax[0].set_xticklabels(["GP - Gabriel Pereira",  "MS - Mousinho da Silveira"])

sns.boxplot(x='school', y='G3_y', data=df_combined, ax=ax[1],color='#F54927', fill=False)
ax[1].set_title ("Resultado final Portugués vs Escuela") #se setea el título del gráfico
ax[1].set(xlabel= 'Escuela', ylabel='Resultado final') #se asignan nombres a las columnas
ax[1].set_xticks([0,1])
ax[1].set_xticklabels(["GP - Gabriel Pereira",  "MS - Mousinho da Silveira"])

plt.tight_layout()

plt.show()
plt.close(fig) 
```

### Interpretación del gráfico

Se observan diferencias en la distribución de las calificaciones entre las escuelas. 
Sin embargo, estas diferencias no pueden considerarse concluyentes sin un análisis estadístico adicional.
Por lo tanto, no es posible afirmar que una escuela tenga mejor rendimiento que otra basándose únicamente en este gráfico.

### Resultados clave

- Las calificaciones parciales (G1 y G2) presentan una fuerte relación con la nota final (G3)
- No se observa una relación clara entre las ausencias y el rendimiento académico
- Las diferencias observadas entre grupos (como género o escuela) no pueden considerarse concluyentes sin validación estadística
- El rendimiento académico parece depender de múltiples factores combinados

### Consideraciones metodológicas

Las conclusiones obtenidas se basan en análisis exploratorios y visualizaciones. 
Por lo tanto, algunas diferencias observadas pueden no ser estadísticamente significativas.

Para validar estas observaciones, sería necesario aplicar pruebas estadísticas (como t-test o ANOVA), evitando interpretaciones basadas únicamente en la inspección visual.

Asimismo, el rendimiento académico puede estar influenciado por variables no incluidas en el dataset, lo que limita el alcance de los resultados.

### Conclusión

El análisis evidencia que el rendimiento académico no depende de una única variable, sino de la interacción de múltiples factores.

Si bien algunas diferencias pueden observarse en los gráficos, estas no siempre resultan significativas, lo que resalta la importancia de complementar el análisis exploratorio con herramientas estadísticas.

Este tipo de enfoque permite obtener conclusiones más robustas y evita interpretaciones simplistas de los datos.
